```python
#duration vs price
import pandas as pd
from scipy.stats import pearsonr

df = pd.read_csv('Clean_Dataset.csv')
df=df.drop(columns='Unnamed: 0')

r, p = pearsonr(df['duration'], df['price'])

print(f"r = {r:.4f}")
print(f"p-value = {p:.6f}")

if p < 0.05:
    print("H0 rejetée : corrélation significative")
else:
    print("H0 non rejetée")
```

avec r = 0.20, cette corrélation reste faible : la durée du vol influence légèrement le prix, mais elle ne l'explique pas à elle seule. D'autres facteurs comme la classe (Business/Economy) ou la compagnie aérienne ont un impact bien plus fort sur le prix du billet.

```python
#test ANOVA 
from scipy.stats import f_oneway

# ANOVA airline vs price
groups = [df[df['airline'] == a]['price'].values for a in df['airline'].unique()]
f, p = f_oneway(*groups)

print(f"F = {f:.2f}")
print(f"p-value = {p:.8f}")

if p < 0.05:
    print("H0 rejetée : différence significative entre les groupes")
else:
    print("H0 non rejetée")
```

Le prix varie significativement selon la compagnie aérienne. 

```python
from scipy.stats import chi2_contingency 

crros_table = pd.crosstab(df['departure_time'], df['stops'])

chi2, p, dof, ex = chi2_contingency(crros_table)

print(f"p-value = {p:.4f}")

if p < 0.05:
    print("les deux variables est depandent ")
else:
    print("les deux variables est indepandent")
```

```python
from scipy.stats import chisquare
# know the variables that you'll work on "airlines","source_city","destination_city" 
# make the H0 hypothesis
# make the test

df.head(10)

## we'll say that 70% of our flights are served by the airline AirAsia (discrete categorical)
## we'll also say that AirAsia does 60% of it's flights are done from delhi to mumbai

# Count observed frequencies
observed_airasia = (df['airline'] == 'AirAsia').sum()
observed_other = len(df) - observed_airasia
observed = [observed_airasia, observed_other]

# Expected frequencies under H0 (70% AirAsia, 30% other)
n = len(df)
expected = [0.7 * n, 0.3 * n]

# Chi-square test
chi2_stat, p_value = chisquare(f_obs=observed, f_exp=expected)
print(f"AirAsia proportion test: chi2 = {chi2_stat:.4f}, p-value = {p_value:.4f}")

# Filter AirAsia flights
airasia_df = df[df['airline'] == 'AirAsia']

# Observed: number of Delhi->Mumbai flights vs. all other routes
observed_del_mum = ((airasia_df['source_city'] == 'Delhi') & 
                    (airasia_df['destination_city'] == 'Mumbai')).sum()
observed_other_routes = len(airasia_df) - observed_del_mum
observed = [observed_del_mum, observed_other_routes]

# Expected under H0: 60% Delhi->Mumbai, 40% other routes 
m = len(airasia_df)
expected = [0.6 * m, 0.4 * m]

# Chi-square test
chi2_stat2, p_value2 = chisquare(f_obs=observed, f_exp=expected)
print(f"Delhi→Mumbai proportion (AirAsia only): chi2 = {chi2_stat2:.4f}, p-value = {p_value2:.4f}")
```
