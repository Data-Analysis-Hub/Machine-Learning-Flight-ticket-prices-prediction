```python
import pandas as pd
from sklearn import utils,preprocessing

df=pd.read_csv('Clean_Dataset.csv')
df=df.drop(columns='Unnamed: 0')
df=utils.shuffle(df,random_state=1) 
```

the columns (features) that should be encoded are:
airline, flights, departure & arrival times, source & destination cities, stops, class

the other ones are already numerical & don't need any more encoding (days_left, price, duration)

i suppose all the data in here is relevant to use in the ML Training

[doc for preprocessing](https://scikit-learn.org/stable/modules/preprocessing.html#encoding-categorical-features)

```python

#### Preprocessing (Encoding) 
categorical_df=df[['airline','source_city','destination_city','days_left','arrival_time','departure_time']]
encoder_cat=preprocessing.OneHotEncoder(sparse_output=False)
array_encoded = encoder_cat.fit_transform(categorical_df)

encoded_cat = pd.DataFrame(
    array_encoded,
    columns = encoder_cat.get_feature_names_out(categorical_df.columns)
)
encoded_cat
```

```python
#### Standard Scaler

numerical_df=df[['duration','price']]
scaler=preprocessing.StandardScaler()
scaled=scaler.fit_transform(numerical_df)
scaled
```

```python
mapping_stops = {
    'zero': 0,
    'one': 1,
    'two_or_more': 2
}

df['stops'] = df['stops'].map(mapping_stops)

print(df['stops'].unique())
```

```python
print(df['class'].unique())
```

```python
mapping_class = {
    'Economy': 0,
    'Business': 1,
   
}

df['class'] = df['class'].map(mapping_class)

print(df['class'].unique())
```
