4️⃣ Multivariate analysis

```python
plt.figure(figsize=(12, 5))
sns.scatterplot(data = df, x = df['duration'], y = df['price'])
plt.title("flight duration vs price")
```

```python
plt.figure(figsize=(12, 5))
sns.barplot(data = df , x="airline", y ="price", errorbar=None, hue = 'airline')
plt.title("airline vs price", c = 'k', fontsize = 20)
```

```python
plt.figure(figsize=(12, 5))
sns.barplot(data = df, x = df['stops'], y = df['price'], errorbar= None, hue = df['stops'])
plt.title("number of stopovers vs. price", c = 'k', fontsize = 20)
```

```python
plt.figure(figsize=(12, 5))
sns.lineplot(data = df , x = df['days_left'], y = df['price'], c = 'orange')
plt.title("days before flight vs price", c = 'k', fontsize = 20)
```
