# 4️⃣ Multivariate analysis

```python
plt.figure(figsize=(12, 5))
sns.regplot(data = df, x = df['duration'], y = df['price'])
plt.title("flight duration vs price")
```

```python
order_by_price =df.groupby('airline') ['price'].mean().sort_values(ascending=False).index

plt.figure(figsize=(12, 5))
sns.barplot(data = df , x="airline", y ="price", errorbar=None, hue = 'airline', order=order_by_price)
plt.title("airline vs price", c = 'k', fontsize = 20)
```

```python
order_list = df.groupby('stops')['price'].mean().sort_values(ascending=False).index

plt.figure(figsize=(12, 5))
sns.barplot(data = df, x = df['stops'], y = df['price'], errorbar= None, hue = df['stops'], order = order_list)
plt.title("number of stops overs vs. price", c = 'k', fontsize = 20)
```

```python
plt.figure(figsize=(12, 5))
sns.lineplot(data = df , x = df['days_left'], y = df['price'], c = 'orange')
plt.title("days before flight vs price", c = 'k', fontsize = 20)
```

```python
pivot = df.pivot_table(
    index="stops",
    columns="airline",
    values="price",
    aggfunc="mean"
)

plt.figure(figsize=(12, 6))

sns.heatmap(
    pivot,
    cmap="coolwarm",
    annot=True,
    fmt=".0f"
)

plt.title("Average Price by Airline and Number of Stops")
plt.xlabel("Airline")
plt.ylabel("Stops")

plt.xticks(rotation=45)
plt.show()
```

```python
pivot2 = df.pivot_table(
    index="departure_time",
    columns="class",
    values="price",
    aggfunc="mean"
)

plt.figure(figsize=(7,4))

sns.heatmap(
    pivot2,
    cmap="coolwarm",
    annot=True,
    fmt=".0f"
)

plt.title("Price by Departure Time and Class")
plt.show()
```

Cette heatmap montre que la classe est le facteur principal influençant le prix des billets, avec des tarifs beaucoup plus élevés en Business qu’en Economy, tandis que l’heure de départ a un impact secondaire légèrement visible sur les prix.
